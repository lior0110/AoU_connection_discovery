## Get the semantic encoder from HuggingFace 🤗

```python

# 2. Load SapBERT model locally
print("\nLoading SapBERT model (cambridgeltl/SapBERT-from-PubMedBERT-fulltext)...")
model = SentenceTransformer("cambridgeltl/SapBERT-from-PubMedBERT-fulltext")

# # 2. Load MedEmbed model locally
# print("\nLoading MedEmbed model (abhinand/MedEmbed-large-v0.1)...")
# model = SentenceTransformer("abhinand/MedEmbed-large-v0.1")
```

## Generate semantic embeddings for laboratory tests names

```python

linearly_correlated_embeddings = {}

for protein in linearly_correlated.index:
    print(f"Embedding {protein}...")
    inx = protein
    if " | " in protein:
        protein = protein.split(" | ")
        print(f"Protein name contains multiple terms: {protein}.")
    linearly_correlated_embeddings[inx] = model.encode(protein, convert_to_tensor=True, convert_to_numpy=True)

# linearly_correlated_embeddings
```

```python

for key, value in linearly_correlated_embeddings.items():
    # print(f"{key}: {value[:5]}...")  # Print the first 5 elements of each embedding
    # print(f"{key}: {value.shape}")  # Print the shape of each embedding
    # print(f"{key}: {len(value)}")  # Print the length of each embedding
    # see how many vectors are in the embedding
    if len(value.shape) > 1:
        print(f"{key}: {value.shape[0]} vectors in the embedding")
    else:
        print(f"{key}: 1 vector in the embedding")

```

```python

monotonically_correlated_embeddings = {}

for protein in monotonically_correlated.index:
    print(f"Embedding {protein}...")
    inx = protein
    if " | " in protein:
        protein = protein.split(" | ")
        print(f"Protein name contains multiple terms: {protein}.")
    monotonically_correlated_embeddings[inx] = model.encode(protein, convert_to_tensor=True, convert_to_numpy=True)

# monotonically_correlated_embeddings
```

```python

non_monotonically_correlated_embeddings = {}

for protein in non_monotonically_correlated.index:
    print(f"Embedding {protein}...")
    inx = protein
    if " | " in protein:
        protein = protein.split(" | ")
        print(f"Protein name contains multiple terms: {protein}.")
    non_monotonically_correlated_embeddings[inx] = model.encode(protein, convert_to_tensor=True, convert_to_numpy=True)

# non_monotonically_correlated_embeddings
```

## find the close tests to each non_monotonically_correlated lab test

```python

protein_dists = {}

for key, value in non_monotonically_correlated_embeddings.items():
    # see how many vectors are in the embedding
    if len(value.shape) > 1:
        print(f"{key}: {value.shape[0]} vectors in the embedding")

        dists_to_target_protein = pd.Series(dtype=float, name="distance_to_target_protein",)
        # get the minmal cosine distance
        for val in value:
            # print(f"Vector shape: {val.shape}")
            # compute cosine distances to linearly_correlated results
            for key2, value2 in linearly_correlated_embeddings.items():
                # see how many vectors are in the embedding
                if len(value2.shape) > 1:
                    print(f"{key2}: {value2.shape[0]} vectors in the embedding")
                    # compute cosine distance between the two embeddings
                    dists = []
                    for v in value2:
                        cosine_distance = np.dot(val, v) / (np.linalg.norm(val) * np.linalg.norm(v))
                        dists.append(np.abs(cosine_distance))
                    if key2 not in dists_to_target_protein:
                        dists_to_target_protein[key2] = np.min(dists)
                    elif key2 in dists_to_target_protein:
                        if np.min(dists) < dists_to_target_protein[key2]:
                            dists_to_target_protein[key2] = np.min(dists)
                else:
                    print(f"{key2}: 1 vector in the embedding")
                    # compute cosine distance between the two embeddings
                    cosine_distance = np.dot(val, value2) / (np.linalg.norm(val) * np.linalg.norm(value2))
                    if key2 not in dists_to_target_protein:
                        dists_to_target_protein[key2] = np.abs(cosine_distance)
                    elif key2 in dists_to_target_protein:
                        if np.abs(cosine_distance) < dists_to_target_protein[key2]:
                            dists_to_target_protein[key2] = np.abs(cosine_distance)

            # compute cosine distances to monotonically_correlated results
            for key2, value2 in monotonically_correlated_embeddings.items():
                # see how many vectors are in the embedding
                if len(value2.shape) > 1:
                    print(f"{key2}: {value2.shape[0]} vectors in the embedding")
                    # compute cosine distance between the two embeddings
                    dists = []
                    for v in value2:
                        cosine_distance = np.dot(val, v) / (np.linalg.norm(val) * np.linalg.norm(v))
                        dists.append(np.abs(cosine_distance))
                    if key2 not in dists_to_target_protein:
                        dists_to_target_protein[key2] = np.min(dists)
                    elif key2 in dists_to_target_protein:
                        if np.min(dists) < dists_to_target_protein[key2]:
                            dists_to_target_protein[key2] = np.min(dists)
                else:
                    print(f"{key2}: 1 vector in the embedding")
                    # compute cosine distance between the two embeddings
                    cosine_distance = np.dot(val, value2) / (np.linalg.norm(val) * np.linalg.norm(value2))
                    if key2 not in dists_to_target_protein:
                        dists_to_target_protein[key2] = np.abs(cosine_distance)
                    elif key2 in dists_to_target_protein:
                        if np.abs(cosine_distance) < dists_to_target_protein[key2]:
                            dists_to_target_protein[key2] = np.abs(cosine_distance)

    else:
        print(f"{key}: 1 vector in the embedding")

        dists_to_target_protein = pd.Series(dtype=float, name="distance_to_target_protein",)
        # compute cosine distances to linearly_correlated results
        for key2, value2 in linearly_correlated_embeddings.items():
            # see how many vectors are in the embedding
            if len(value2.shape) > 1:
                print(f"{key2}: {value2.shape[0]} vectors in the embedding")
                # compute cosine distance between the two embeddings
                dists = []
                for v in value2:
                    cosine_distance = np.dot(value, v) / (np.linalg.norm(value) * np.linalg.norm(v))
                    dists.append(np.abs(cosine_distance))
                if key2 not in dists_to_target_protein:
                    dists_to_target_protein[key2] = np.min(dists)
                elif key2 in dists_to_target_protein:
                    if np.min(dists) < dists_to_target_protein[key2]:
                        dists_to_target_protein[key2] = np.min(dists)
            else:
                print(f"{key2}: 1 vector in the embedding")
                # compute cosine distance between the two embeddings
                cosine_distance = np.dot(value, value2) / (np.linalg.norm(value) * np.linalg.norm(value2))
                if key2 not in dists_to_target_protein:
                    dists_to_target_protein[key2] = np.abs(cosine_distance)
                elif key2 in dists_to_target_protein:
                    if np.abs(cosine_distance) < dists_to_target_protein[key2]:
                        dists_to_target_protein[key2] = np.abs(cosine_distance)

        # compute cosine distances to monotonically_correlated results
        for key2, value2 in monotonically_correlated_embeddings.items():
            # see how many vectors are in the embedding
            if len(value2.shape) > 1:
                print(f"{key2}: {value2.shape[0]} vectors in the embedding")
                # compute cosine distance between the two embeddings
                dists = []
                for v in value2:
                    cosine_distance = np.dot(value, v) / (np.linalg.norm(value) * np.linalg.norm(v))
                    dists.append(np.abs(cosine_distance))
                if key2 not in dists_to_target_protein:
                    dists_to_target_protein[key2] = np.min(dists)
                elif key2 in dists_to_target_protein:
                    if np.min(dists) < dists_to_target_protein[key2]:
                        dists_to_target_protein[key2] = np.min(dists)
            else:
                print(f"{key2}: 1 vector in the embedding")
                # compute cosine distance between the two embeddings
                cosine_distance = np.dot(value, value2) / (np.linalg.norm(value) * np.linalg.norm(value2))
                if key2 not in dists_to_target_protein:
                    dists_to_target_protein[key2] = np.abs(cosine_distance)
                elif key2 in dists_to_target_protein:
                    if np.abs(cosine_distance) < dists_to_target_protein[key2]:
                        dists_to_target_protein[key2] = np.abs(cosine_distance)

    protein_dists[key] = dists_to_target_protein.sort_values(ascending=True)

```
